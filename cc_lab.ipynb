```python
#question 1
roll_no = "1024170238"
print("the roll no. is : " , roll_no)
L = [int(digit)*10 for digit in roll_no]
print(L)
L.append(27)
print("L after appending 27 : " , L)
L.insert(3,9)
print("L after inserting 9 at 3rd index :" , L)
L.remove(0)
print("L after removing first 0 from the list : " , L)
L.pop(5)
print("L after deleting at 5th index : " , L)
L.sort()
print("After sorting : " , L)
L.sort(reverse=True)
print("After reversing the sorted list : " , L)
print("First three elements : " , L[:3])
print("Last three elements : " , L[-3:])
sum_of_list = sum(L)
length = len(L)
avg = sum_of_list/length
more_than_avg = [i for i in L if i>avg]
print("average : " , avg)
print("greater than average : " , more_than_avg)
```
