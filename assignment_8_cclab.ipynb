```python
import re 
import nltk 
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords 
```

```python
text = 'sports is an important part of our life. It keeps us fit physically and helps us mentally active. Playing sports teaches teamwork, discipline, and patience. Cricket, football, and badminton are popular sports. Sports also help us reduce stress and make new friends. Everyone should play sports to stay healthy.'
```

```python
text = text.lower()
text = re.sub(r'[^\w\s]', '', text)
print("Text after 1st step :\n", text)
```

```python
print('word tokens : ' , word_tokenize(text))
print('sentences tokens : ' , sent_tokenize(text))
```

```python
print('Word tokens: ' , word_tokenize(text))
print('split function : ' , text.split())
```

```python
words = word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered_words = []
for word in words:
    if word not in stop_words:
        filtered_words.append(word)
print('after removing stopwords : ' , filtered_words)
```

```python
from nltk.probability import FreqDist
```

```python
freq = FreqDist(filtered_words)
print('frequency is : ' , freq)
```

```python
from nltk.stem import PorterStemmer , WordNetLemmatizer
```

```python
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print(words)
```

```python
words = word_tokenize(text)
stop_words = set(stopwords.words('english'))
filtered_words = []
for word in words:
    if word not in stop_words:
        filtered_words.append(word)
print('after removing stopwords : ' , filtered_words)
```

```python
stemmed_words =[] 
ps = PorterStemmer()
for word in filtered_words:
    stemmed_words.append(ps.stem(word))
print('After stemming : ' , stemmed_words)
```

```python
lm = WordNetLemmatizer()
lm_word = [] 
for word in filtered_words: 
    lm_word.append(lm.lemmatize(word))
print('After lemmatization : ', lm_word)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
c_vect = CountVectorizer()
texts = [
    "Sports improve fitness and health.",
    "Football and cricket are popular sports.",
    "Technology improves health and education."
]
bow = c_vect.fit_transform(texts)
print(c_vect.get_feature_names_out())
print(bow.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(texts)
print(tfidf.get_feature_names_out())
print(X.toarray())
```

```python
features = tfidf.get_feature_names_out()
for i in range(len(texts)):
    scores = X.toarray()[i]
    top3 = scores.argsort()[-3:][::-1]
    print("Text:", texts[i])
    print("Top 3 keywords:")
    for index in top3:
        print(features[index], ":", round(scores[index], 3))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
text1 = 'Artificial intelligence helps machines learn from data. AI is used in healthcare and education. It can recognize images and understand language. AI improves automation and decision making.'
text2 = 'Blockchain stores data in secure digital blocks. It is used in banking and cryptocurrency. Blockchain provides transparency and security. It helps maintain reliable digital records.'
text1 = re.sub(r'[^\w\s]', '', text1.lower())
text2 = re.sub(r'[^\w\s]', '', text2.lower())
words1 = text1.split()
words2 = text2.split()
print("Tokens for text1:", words1)
print("Tokens for text2:", words2)
set1 = set(words1)
set2 = set(words2)
jaccard = len(set1.intersection(set2)) / len(set1.union(set2))
print("\nJaccard Similarity:", round(jaccard, 3))

text_arr = [text1, text2]
vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(text_arr)

cosine = cosine_similarity(tfidf[0:1], tfidf[1:2])[0][0]
print("Cosine Similarity:", round(cosine, 3))
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt
reviews = [
    "This phone has an amazing camera and excellent battery life.",
    "The product is terrible and its quality is very poor.",
    "The phone is okay and works as expected.",
    "I love this phone because it is fast and easy to use.",
    "The battery is disappointing and the phone is very slow."
]
positive_reviews = []

for review in reviews:
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity
    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"
    print("\nReview:", review)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("Sentiment:", sentiment)

text = " ".join(positive_reviews)
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(text)
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```
