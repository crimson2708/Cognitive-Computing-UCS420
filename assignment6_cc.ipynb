```python
#ques 1 
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_temperature = temperature + 2
print(corrected_temperature)
fahrenheit = (9/5) * corrected_temperature + 32
print(fahrenheit)
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print(greater_than_32)
count = np.sum(corrected_temperature > 32)
print(count)
```

```python
#ques 2 
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
total_steps = np.sum(steps)
print(total_steps)
mean_steps = np.mean(steps)
print(mean_steps)
maximum = np.max(steps)
minimum = np.min(steps)
print(maximum)
print(minimum)
daily_total = np.sum(steps, axis=0)
print(daily_total)
user_total = np.sum(steps, axis=1)
print(user_total)
max_index = np.argmax(steps)
max_position = np.unravel_index(max_index, steps.shape)
print(max_position)
print(steps[max_position])
```

```python
#ques 3 

original = np.array([1, 2, 3, 4, 5, 6])
print(original)

subset = original[1:5]
print(subset)

subset[0] = 999
print(original)
print(subset)
copied = original[1:5].copy()
copied[0] = 500
print(original)
print(copied)
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)

first_row = matrix[0, :]
print(first_row)

last_row = matrix[-1, :]
print(last_row)

second_column = matrix[:, 1]
print(second_column)

selected = matrix[0:2, 1:3]
print(selected)

flattened = matrix.flatten()
raveled = matrix.ravel()

print(flattened)
print(raveled)
raveled[0] = 100
print(raveled)
print(matrix)

flattened[1] = 200
print(flattened)
print(matrix)
print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

```python
#ques 4 
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
print(X.shape)
print(X.ndim)
XT = X.T
print(XT)
XT_X = XT @ X
print(XT_X)
inverse = np.linalg.inv(XT_X)
print(inverse)
beta = inverse @ XT @ y
print(beta)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print(predicted_score)
```
