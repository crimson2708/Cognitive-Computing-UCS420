```python
#ques 1
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

```python
#ques 2
arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]
print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]])
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
#ques4 
your_name = np.linspace(10, 100, 25)
print("Array:", your_name)
print("Dimensions:", your_name.ndim)
print("Shape:", your_name.shape)
print("Total elements:", your_name.size)
print("Data type:", your_name.dtype)
print("Total bytes:", your_name.nbytes)
transpose_array = your_name.reshape(25, 1)
print("Transpose using reshape():")
print(transpose_array)
print("Transpose using T:")
print(your_name.T)
print("Can T perform the same operation for a 1-D array?", your_name.T.shape == your_name.shape)
```

```python
#ques 5 
ucs420_your_name = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]])
print("Original array:")
print(ucs420_your_name)
print("Mean:", np.mean(ucs420_your_name))
print("Median:", np.median(ucs420_your_name))
print("Maximum:", np.max(ucs420_your_name))
print("Minimum:", np.min(ucs420_your_name))
print("Unique elements:", np.unique(ucs420_your_name))
reshaped_ucs420_your_name = ucs420_your_name.reshape(4, 3)
print("Reshaped array:")
print(reshaped_ucs420_your_name)
resized_ucs420_your_name = np.resize(ucs420_your_name, (2, 3))
print("Resized array:")
print(resized_ucs420_your_name)
```
